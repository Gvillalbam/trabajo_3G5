```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Creamos el dataset con la información proporcionada
data = {
    'timestamp': pd.date_range(start='2023-01-01', periods=5, freq='H'),
    'Archivo': ['file1.exe', 'report.docx', 'script.js', 'img1.png', 'setup.exe'],
    'Tamaño Extension': ['150 KB', '50 KB', '25 KB', '200 KB', '1 MB'],
    'Hash_MD5': ['f5e0b56b9b12f053', '2c6e3953cd092276', '45fbd139f9d72a3b', '8f8fbbc51c508cb1', 'd1e9a738b41adacb'],
    'Hash_SHA256': ['3d4d5edb9730a6ec5fa924d08fc60c69', 'b9535a79964bb8399ac44d075e9cb17c', '3e213db71c973f67425f4b467ec7ab0e', '9700d4ef2fc92835b7dbde3958d13cd7', '3e5512db81c973f67425f4b467ec7ab0e'],
    'Comportamiento': ['Ejecución sospechosa', 'Comportamiento normal', 'Ejecuta comandos remotos', 'Comportamiento normal', 'Ejecución sospechosa'],
    'Etiqueta_Malware': ['Malicioso', 'No malicioso', 'Malicioso', 'No malicioso', 'Malicioso']
}

# Creamos el DataFrame
df = pd.DataFrame(data)

print("1. Explorando el DataFrame:")
print(df.head())
print("\nInformación del DataFrame:")
print(df.info())
print("\nEstadísticas descriptivas:")
print(df.describe())

# 2. Selección de datos
print("\n2. Selección de datos:")
# Seleccionar una columna (devuelve una Serie)
print("\nSelección de una columna (Serie):")
print(df['Archivo'])

# Selección de múltiples columnas (devuelve un DataFrame)
print("\nSelección de múltiples columnas (DataFrame):")
print(df[['Archivo', 'Tamaño Extension', 'Comportamiento']])

# Selección por índice con iloc
print("\nSelección por índice con iloc (primeras 2 filas, columnas 0 y 2):")
print(df.iloc[0:2, [0, 2]])

# Selección por nombre con loc
print("\nSelección por nombre con loc (filas donde comportamiento es 'Ejecución sospechosa'):") 
print(df.loc[df['Comportamiento'] == 'Ejecución sospechosa', ['timestamp', 'Archivo', 'Comportamiento']])

# 3. Filtrado de datos
print("\n3. Filtrado de datos:")
# Filtrar archivos maliciosos
archivos_maliciosos = df[df['Etiqueta_Malware'] == 'Malicioso']
print("\nArchivos maliciosos:")
print(archivos_maliciosos[['timestamp', 'Archivo', 'Comportamiento']])

# Filtrado múltiple (archivos sospechosos y maliciosos)
sospechosos_maliciosos = df[(df['Comportamiento'] == 'Ejecución sospechosa') & (df['Etiqueta_Malware'] == 'Malicioso')]
print("\nArchivos sospechosos y maliciosos:")
print(sospechosos_maliciosos)

# 4. Transformación de datos
print("\n4. Transformación de datos:")
# Convertir la columna 'Tamaño Extension' a una unidad numérica
def convertir_tamaño(tamaño):
    if 'KB' in tamaño:
        return int(tamaño.split(' ')[0]) * 1024  # KB a bytes
    elif 'MB' in tamaño:
        return int(tamaño.split(' ')[0]) * 1024 * 1024  # MB a bytes
    return 0

df['Tamaño_Bytes'] = df['Tamaño Extension'].apply(convertir_tamaño)
print("\nDataFrame con nueva columna 'Tamaño_Bytes' en bytes:")
print(df[['Archivo', 'Tamaño Extension', 'Tamaño_Bytes']])

# 5. Agregación de datos
print("\n5. Agregación de datos:")
# Agrupar por comportamiento y contar archivos
conteo_por_comportamiento = df.groupby('Comportamiento').size()
print("\nConteo de archivos por comportamiento:")
print(conteo_por_comportamiento)

# 6. Pivotado de datos
print("\n6. Pivotado de datos:")
# Tabla pivote de tamaño total por comportamiento y etiqueta de malware
pivot = pd.pivot_table(df, 
                       values='Tamaño_Bytes', 
                       index='Comportamiento', 
                       columns='Etiqueta_Malware', 
                       aggfunc='sum')
print("\nTabla pivote de tamaño total por comportamiento y etiqueta de malware:")
print(pivot)

# 7. Detección de anomalías simples
print("\n7. Detección de anomalías simples:")
# Calcular el Z-score para identificar archivos de tamaño atípico
df['z_score'] = (df['Tamaño_Bytes'] - df['Tamaño_Bytes'].mean()) / df['Tamaño_Bytes'].std()
anomalias = df[abs(df['z_score']) > 1.5]  # Consideramos como anómalos Z-scores > 1.5
print("\nPosibles archivos anómalos por tamaño:")
print(anomalias[['timestamp', 'Archivo', 'Comportamiento', 'Tamaño_Bytes', 'z_score']])
```
